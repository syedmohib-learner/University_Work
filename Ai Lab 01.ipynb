### Exercise: Dir and Helo

```python
s = "Mohib"
print("Methods available for string:")
print(dir(s))
print("\nHelp for find() method:")
help(s.find)
print("\n--- String Methods ---")
print("Original string:", s)
print("Upper:", s.upper())
print("Lower:", s.lower())
print("Capitalize:", s.capitalize())
print("Count of 'o':", s.count('o'))
print("Find 'h':", s.find('h'))
print("Find 'x':", s.find('x'))
print("Starts with 'M':", s.startswith('M'))
print("Ends with 'b':", s.endswith('b'))
print("Replace 'o' with 'a':", s.replace('o', 'a'))
print("Is alphabetic:", s.isalpha())
print("Is digit:", s.isdigit())
print("Is lowercase:", s.islower())
print("Is uppercase:", s.isupper())
print("Length of string:", len(s))
```

### (i)Write a Python program to swap 4 variables values (input four values.)


```python
a, b, c, d = 5, 2, 7, 9
print("Before swapping")
print(f"a={a}, b={b}, c={c}, d={d}")
a, b, c, d = d, c, b, a
print("After swapping")
print(f"a={a}, b={b}, c={c}, d={d}")
```

### (ii) Write a Python program to convert temperatures to and from Celsius,Fahrenheit.

```python
celsius = 60
fahrenheit = (celsius * 9 / 5) + 32
print(f"Enter temp in Celsius: {celsius}°C")
print(f"Temperature in Fahrenheit is: {fahrenheit:.0f}")
```

### Exercise List (i)

```python
fruits = ['apple', 'banana', 'mango']
print("Original List:", fruits)
fruits.append('orange')
print("After append('orange'):", fruits)
print("Count of 'banana':", fruits.count('banana'))
fruits.extend(['grapes', 'watermelon'])
print("After extend(['grapes', 'watermelon']):", fruits)
print("Index of 'mango':", fruits.index('mango'))
fruits.insert(1, 'guava')
print("After insert(1, 'guava'):", fruits)
removed = fruits.pop()
print("Popped fruit:", removed)
print("After pop():", fruits)
fruits.remove('guava')
print("After remove('guava'):", fruits)
fruits.reverse()
print("After reverse():", fruits)
fruits.sort()
print("After sort():", fruits)
```

### Exercise List (ii):

```python
def count_matching_ends(lst):
    count = 0
    for s in lst:
        if len(s) >= 2 and s[0] == s[-1]:
            count += 1
    return count
sample_list = ['abc', 'xyz', 'aba', '1221']
print(count_matching_ends(sample_list))
```

### Exercise: List Comprehensions (i)

```python
words = ['Python', 'AI', 'Programming', 'Lab', 'University', 'Code']
lowercased = [w.lower() for w in words if len(w) > 5]
print(lowercased)
```

### Exercise: List Comprehensions (ii)

```python
color_list = ['Red', 'Green', 'White', 'Black', 'Pink', 'Yellow']
filtered = [color_list[i] for i in range(len(color_list)) if i not in (0, 4, 5)]
print(filtered)
```

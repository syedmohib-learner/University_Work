```python
height = float(input("Enter height: "))
width = float(input("Enter width: "))
depth = float(input("Enter depth: "))
volume = height * width * depth
print(f"Volume of the cube = {volume} cm^3")
if volume >= 1 and volume <= 10:
    label = "Extra Small"
elif volume >= 11 and volume <= 25:
    label = "Small"
elif volume >= 26 and volume <= 75:
    label = "Medium"
elif volume >= 76 and volume <= 100:
    label = "Large"
elif volume >= 101 and volume <= 250:
    label = "Extra Large"
else:
    label = "Extra-Extra Large"
print(f"Label: {label}")
```

### Question 2(ii):

```python
clist = ['Canada', 'USA', 'Mexico', 'Australia']
# List the countries
print("Countries in the list:")
for country in clist:
    print(country)
# 1. Loop that counts from 0 to 100
print("\n1. Counting from 0 to 100:")
for i in range(0, 101):
    print(i, end=" ")
print()
# 2. Multiplication table using a loop
print("\n2. Multiplication table (for number 5):")
number = 5
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
# 3. Output numbers 1 to 10 backwards using a loop
print("\n3. Numbers 1 to 10 backwards:")
for i in range(10, 0, -1):
    print(i, end=" ")
print()
# 4. Loop that counts all even numbers up to 10
print("\n4. Even numbers up to 10:")
for i in range(0, 11):
    if i % 2 == 0:
        print(i, end=" ")
print()
# 5. Loop that sums the numbers from 100 to 200
print("\n5. Sum of numbers from 100 to 200:")
total = 0
for i in range(100, 201):
    total += i
print(f"Sum = {total}")
```

### Question 2(iii)-2

```python
# A for loop is used when the number of iterations is known in advance, or when iterating directly over a sequence (list, string, range, etc.). A while loop is used when the number of iterations is not known ahead of time and depends on a condition remaining True — it keeps running until that condition becomes False.
```

### Question 2(iii)-3

```python
total = 0
i = 1
while i<=5:
    total += i
    i += 1
print(f"Sum = {total}")
# Yes, a while loop can accumulate a sum just like a for loop, as shown above.
```

### Exercise 1

#### 1.I a Python program to square and cube every number in a given list of integers using Lambda. 

```python
numbers=[1,2,3,4]
square = list(map(lambda x: x**2,numbers))
cube = list(map(lambda x: x**3, numbers))
print("Original list:",numbers)
print("Square of every number is:",square)
print("Cube f every number is:",cube)
```

#### 1.II a Python program to find if a given string starts with a given character using Lambda. 

```python
string = input("Enter a string: ")
char = input("Enter a character to check: ")
starts_with = lambda s, c: s.startswith(c)
if starts_with(string, char):
    print(f'"{string}" starts with "{char}" -> True')
else:
    print(f'"{string}" starts with "{char}" -> False')
```

#### 1.III a Python program to extract year, month, date and time using Lambda. 

```python
import datetime
now = datetime.datetime.now()
get_year = lambda dt: dt.year
get_month = lambda dt: dt.month
get_date = lambda dt: dt.strftime("%Y-%m-%d")
get_time = lambda dt: dt.strftime("%H:%M:%S")
print("Current DateTime:", now)
print("Year:", get_year(now))
print("Month:", get_month(now))
print("Date:", get_date(now))
print("Time:", get_time(now))
```

### Exercise 2

#### 2.I You have collected information about cities in your province. You decide to store each city’s name, population, and mayor in a file. Write a python program to accept the data for a number of cities from the keyboard and store the data in a file in the order in which they’re entered

```python
n = int(input("How many cities do you want to enter? "))
with open("cities.txt", "w") as f:
    for i in range(n):
        print(f"\nCity #{i + 1}")
        name = input("  City name: ")
        population = input("  Population: ")
        mayor = input("  Mayor: ")
        f.write(f"{name},{population},{mayor}\n")
print("\nData successfully saved to 'cities.txt'")
print("\nContents of cities.txt:")
with open("cities.txt", "r") as f:
    print(f.read())
```

```python
import math
import time
import glob
import random

# 1. math module
print("[math module]")
print("math.sqrt(9) =", math.sqrt(9))
print("math.pi =", math.pi)

from math import sqrt
print("Using 'from math import sqrt' -> sqrt(16) =", sqrt(16))

import math as maths
print("Using 'import math as maths' -> maths.sqrt(25) =", maths.sqrt(25))

from math import sqrt as squareroot
print("Using 'from math import sqrt as squareroot' -> squareroot(36) =",
      squareroot(36))

# 2. time module
print("\n[time module]")
current_time = time.time()
print("time.time() ->", current_time)
print("time.ctime() ->", time.ctime())
print("Sleeping for 1 second...")
time.sleep(1)
print("Done sleeping.")

# 3. glob module
print("\n[glob module]")
print("All files in current directory (glob.glob('*')):")
print(glob.glob("*"))
print("All .py files (glob.glob('*.py')):")
print(glob.glob("*.py"))

# 4. random module
print("\n[random module]")
print("random.randint(1, 10) ->", random.randint(1, 10))
print("random.random() ->", random.random())

x = [1, 2, 3, 4, 5]
random.shuffle(x)
print("random.shuffle([1,2,3,4,5]) ->", x)

y = [1, 2, 3, 4, 5]
print("random.sample([1,2,3,4,5], 2) ->", random.sample(y, 2))
```
